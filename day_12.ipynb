```python
# **kwargs

def show_kwargs(**kwargs):
    print(kwargs)
show_kwargs(name = "rock" , age = 20)
```

```python
# print all keyword argument

def show_kwargs(**kwargs):
    print(kwargs)
show_kwargs(city = "pune" , rain = False)
```

```python
def student_info(**record): # pass : same way key and value 
    print(record)
student_info(name = "Rock" , city = "NYC" , age = 20, Grade = "A", sport = "Fighting")
```

```python
# loop the key-value pairs

def loop_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(key, value)
loop_kwargs(name = "rock" , city = "NYC")
```

```python
def multiplication (*nums) :
    result = 1

    for i in nums :
        result *= i

    return result

multiplication(10,20,30)
```

```python
def addition (*nums) :
    result = 0

    for i in nums :
        result += i

    return result

addition(10,20,30)
```

```python
def division (*nums) :
    result = 0

    for i in nums :
        result /= i

    return result 

division(10,2)
```

```python
def substaction (*nums) :
    result = 1

    for i in nums :
        result -= i

    return result

substaction(50,10,10)
```
