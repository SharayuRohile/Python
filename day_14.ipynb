```python
# Lambda Function : Anonymous function
# In single line we can write code
# when we wanted a quick code
```

```python
lambda a , b : a + b
```

```python
# Find Addition of two numbers

add = lambda a , b : a + b
add (19 , 90)
```

```python
 # Find Multiplication of two numbers

multiply = lambda a , b : a * b
multiply(10 , 10)
```

```python
# Find Subtraction of two numbers

subtraction = lambda a , b : a - b
subtraction(90 , 10)
```

```python
# Print Hello World

name = lambda : "hello world"
name()
```

```python
# Print Good Morning

greet = lambda : "Good Morning"
greet()
```

```python
def greet () :
    return "Good Morning"

greet()
```

```python
# Find the square of a number

square = lambda a  : a * a
square(5)
```

```python
# Find the square of a number using def

def square (a) :
    return a * a

square(9)
```

```python
# Ckeck if the number is even or odd by def function

def num_check (num) :
    if num % 2 == 0 :
        print("even")

    else :
        ("odd")

num_check(10)
```

```python
# Ckeck if the number is even or odd

even_odd = lambda num : "Even" if num % 2 == 0 else "Odd"
even_odd(6)
```

```python
# Check If you can apply for voting card or not

voting_card_check = lambda age : "can apply" if age >= 18 else "cant apply"
voting_card_check(19)
```

```python
# Convert Uppercase to Lowercase

smallcase = lambda x : x.lower()
smallcase("HELLO")
```

```python
# Check if the string is empty or not

empty = lambda x : "String Empty" if x == "" else "Not Empty"
empty("")

```

```python
# Check if the number is positive or negative

positive_negative = lambda num : "Positive" if num >= 0 else "Negative"
positive_negative(9)
```
