# .append() - Adding elements in last

```python
trip=["Sharayu","Archit","Navin","Aish","Avanti"]
trip.append("Sampada")
print(trip)
```

```python
# Empty List
a=[]
print(a)
```

```python
a.append("car")
print(a)
```

# .extend - for adding two list in one

```python
a=["milk","sugar","cookies"]
b=["rice","wheat","jowar"]
a.extend(b)
print(a)
```

# .insert() - adding element anywhere by index position

```python
names=["Sharayu","Archit","Navin","Aish","Avanti"]
names.insert(1,"Virat")
names.insert(5,"Jay")
print(names)
```

# .remove() - to remove an element

```python
names=["Sharayu","Archit","Navin","Aish","Avanti"]
names.remove("Aish")
print(names)
```

# pop - to remove element from the last and if we want to remove specific element then we use index position

```python
cart=["tshirt","Jeans","Cap","Umbrella"]
cart.pop()
print(cart)
```

```python
cart=["tshirt","Jeans","Cap","Umbrella"]
cart.pop(0)
print(cart)
```

# .clear() - when we want empty list then we used clear

```python
a=["Orange","Blue","Pink","Red"]
a.clear()
print(a)
```

# .index() - it tells the index position of an element . If we have one element more than one time at that time we index of first occurence

```python
a=[1,2,3,4,5,6]
a.index(6)
```

```python
a=[1,8,10,2,3,4,5,6,7,6]
a.index(6)
```

# .count() - it will count the item frequency means how many times that element is present in the list

```python
a=[1,2,3,4,5,6,7,6]
a.count(6)
```

# .sort() - it will sort elements in ascending order

```python
a=[1,8,10,2,3,4,5,6,7,6]
a.sort()
print(a)
```

# .sort(reverse) - it will sort elements in descending order

```python
a=[1,8,10,2,3,4,5,6,7,6]
a.sort(reverse=True)
print(a)
```
