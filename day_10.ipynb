```python
# Dictionary
# It is key : value pair together called items
# key should be unique
# Dictionary is mutable

a={} # Empty dict
```

```python
# Key should be unique if we take same key two times then only one get print
# Unordered

student={"Name" : "Rock" , "Name" : "John"}
print(student)
```

```python
# Add

student={ "Name" : "Rock" , "City" : "Pune" , "Class" : 10 , "Marks" : [88,99,100] }
print(student)
```

```python
# When we have to print only keys
# .keys()

student.keys()
```

```python
# If we want to print only values
# .values()

student.values()
```

```python
student={"Name" : "Sharayu" , "Class" : 10 , "Marks" : [88,78,98] , "Hobbies" : ["Dance","Cooking","Craft"]}
print(student)
```

```python
# .items()

student.items()
```

```python
student_profile = {"Name" :"Sharayu" , 
                   "Class" : 12 , 
                   "Marks" : [99,98,97] , 
                   "City" : "Yavatmal" , 
                   "Sports" : ["Badminton" , "Cricket"]}
print(student_profile)
```

```python
# For loop in Dictionary

for i , j in student_profile.items() :
    print("->" ,i," : " ,j)
```

```python
# When we want to exttract One element

student_profile = {"Name" :"Sharayu" , 
                   "Class" : 12 , 
                   "Marks" : [99,98,97] , 
                   "City" : "Yavatmal" , 
                   "Sports" : ["Badminton" , "Cricket"]}
student_profile["Marks"]
student_profile["City"]
```

```python
# get() when key is not present then also it will not show error

student_profile.get("City")
```

```python
# when key is not present then also it will not show error
# get()
student_profile.get("id" , " Key not present ")
```

```python
# When we call like this it will show error

student_profile["id"]
```

```python
for i , j in student_profile.items() :
    print( i , j )
```

```python
# Update
# We can update our dict

student_profile.update({"Id" : 1 , "Gender" : "Female"})
student_profile.items()
```

```python
for i , j in student_profile.items() :
    print(i, " : " ,j)
```

```python
student_profile.update({"Name" : "Rohit" })
print(student_profile)
```

```python
# Pop
# clear

student_profile.clear()
print(student_profile)
```

```python
a={"Food" : "Oats" , "Drink" : "Coffee"}
b=a.copy()
print(b)
```

```python
# Dictionary of students and marks

marks={"Pradip" : 45,
       "Gaurav" : 78,
       "Sneha" : 90,
       "Ravi" : 32,
       "Meena" : 55}

for k,v in marks.items():
    if v > 50 :
        print(k," = " ,v)
```
