```python
# Exception Handling
```

```python
try:
    num = input()
    result = num / 10
    print(result)

except Exception as e :
    print(e)
```

```python
try:
    my_list = [1,2,3,4]
    my_list[4]

except Exception :
    print("Please check the list index length")
```

```python
# ZeroDivisionError

try:
    10 / 0

except ZeroDivisionError as z :
    print(z)
```

```python
# ValueError

try:
    number = int("abc")

except ValueError as v :
    print(v)
```

```python
# FileNotFoundError

import pandas as pd

try:
    df = pd.read_csv("xyz.csv")

except FileNotFoundError as f :
    print(f)
```

```python
# IndexError

try:
    my_list = [1,2,3,4,5,6,7,8,9]
    my_list[10]

except IndexError as i :
    print(i)
```

```python
# KeyError

try:
    a = {"a" : 100 , "b" : 200}
    a["c"]

except KeyError as k :
    print("Please enter the key which is present in dictionary " , k)
```

```python
# Regex (Regular Expression)
# It is a way to search , match , or manipulate text using patterns
# It is a like a smart search tool
```

```python
import re
```

```python

# \d : Matches any digit (0–9)

text = "My phone number is 12345"

# Search for digits
pattern = r"\d+"    # \d+ : finds all number in the text

#
result = re.findall(pattern,text)
print(result)
```

```python
# \D : Matches any non-digit character

text = "My phone number is 12345"

# Search for digits
pattern = r"\D+"    

#
result = re.findall(pattern,text)
print(result)
```

Common Regex Notations in Python
\d → Matches any digit (0–9)
\D → Matches any non-digit character
\w → Matches any word character (letters, digits, underscore)
\W → Matches any non-word character
\s → Matches any whitespace (space, tab, newline)
\S → Matches any non-whitespace character
. → Matches any single character (except newline)
^ → Matches the start of a string
$ → Matches the end of a string
+ → Matches one or more repetitions
***** → Matches zero or more repetitions
? → Matches zero or one occurrence
{n} → Matches exactly n repetitions
{n,m} → Matches between n and m repetitions
| → OR operator (choice between patterns)
[] → Matches any single character inside the brackets
[^ ] → Matches any character not inside the brackets


```python
# \w : Matches any word character (letters, digits, underscore)

text = "Python_3 is very powerful tool !"

# Search for digits
pattern = r"\w+"    

result = re.findall(pattern,text)
print(result)
```

```python
# \W : Matches any non-word character

text = "Python_3 is great !!!"
pattern = r"\W+"    

result = re.findall(pattern,text)
print(result)
```

```python
# \s : Matches any whitespace (space, tab, newline)

text = "Python_3    is great !!!"
pattern = r"\s+"    

result = re.findall(pattern,text)
print(result)
```

```python
# \S : Matches any non-whitespace character

text = "Python_3 is great !!!"
pattern = r"\S+"    

result = re.findall(pattern,text)
print(result)
```

```python
# . : Matches any single character (except newline)

text = "Python_3 is a great !!!"
pattern = r"."    

result = re.findall(pattern,text)
print(result)
```

```python
# ^ : Matches the start of a string

text = "Python is powerful and fun"
pattern = r"^Python"    

result = re.findall(pattern,text)
print(result)
```

```python
# $ : Matches the end of a string

text = "Python is powerful and fun"
pattern = r"fun$"    

result = re.findall(pattern,text)
print(result)

```

```python
# + : Matches one or more repetitions

text = "rrr rocks really right !"
pattern = r"r+"    

result = re.findall(pattern,text)
print(result)
```
