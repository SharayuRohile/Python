```python
# OOPS - Object Oriented Programming
# Defination - It is a programming paradigm that organizes software design around objects .
# Objects represent entities that combine both data(attributes) and behaviour(methods) .
# Class - Acts as a blueprint that defines atrributes and methods
# An object is an instance created from a class
# Modularity , Reusability , Maintainability , Real world problems
```

```python
# Four Pillars of OOPs are :
# Inheritance
# Abstraction
# Polymorphism
# Encapsulation
```

```python
# Create a class

class Car : 
    def __init__ (self , brand , model , year) : # __init__ is constructor
        self.brand = brand # instance variable   # Self for establishing connection
        self.model = model # instance variable
        self.year = year   # instance variable

car1 = Car("Tata" ,"Punch" ,"2025")  # Object

print(car1.brand)
print(car1.model)
print(car1.year)
```

```python
# Craete a class Bank

class Bank :
    def __init__ (self , bank_name , branch) :
        self.bank_name = bank_name
        self.brach = branch


    # Method

    def info(self) :
        return f"Bank is : {self.bank_name}"

# create an object

bank1 = Bank("SBI" , "Main")

# Calling the attributes

print(bank1.bank_name)
print(bank1.brach)

bank1.info()
```

```python
class ATM:
    def __init__(self, bank_name, location, atm_id):
        self.bank_name = bank_name
        self.location = location
        self.atm_id = atm_id
    
    def info(self):
        return f"ATM {self.atm_id} of {self.bank_name} located at {self.location}"

# Creating an object
atm1 = ATM("State Bank of Python", "Pune", "ATM101")

# Using the method
print(atm1.info())
```
