```python
# Inheritance → Mechanism to create new classes from existing ones, reusing functionality
# It allows child class to inherit properties of parent class
```

```python
class Vehicle :
    def __init__ (self , brand) : # Constructor
        self.brand = brand        # Instance variable

    # Method
    def info(self) :
        return f"Brand name is {self.brand}"

veh1=Vehicle("Tata")

class car(Vehicle) :             # Child class
    def __init__ (self , brand , model) :
        # call parent constructor
        super().__init__(brand)  # To call parent class constructor or method super is used
        self.model = model

    # Polymorphism 
    
    def info(self) :
        return f"Brand is {self.brand} and model is {self.model}"

    # Method
    # Parents class methods can call in child class but child class method cant call in parent class

    def car_info(self) :
        return f"brand is {self.brand} and model is {self.model}"

new_car = car("Tata" , "Nexon") # Object Creation

new_car.info()         # Calling Method

# print(new_car.model) # When we want to call model 
    
```

```python
# inheritance
# parent class
# info : return - bank name

class Bank:
    def __init__(self,bank_name):
        self.bank_name = bank_name

    # method
    def info(self):
        return f"Bank Name : {self.bank_name}"

# child class
class User(Bank):
    def __init__(self, bank_name , user_name):
        # Call parent constructor
        super().__init__(bank_name)
        self.user_name = user_name

    # Method : Poly
    def info(self):
        return f"Bank Name : {self.bank_name} and It is the Nice Bank"

    def user_info(self):
        return f"User: {self.user_name}  : Bank  {self.bank_name}"
```

```python
# creating objects
bank_1 = Bank("State Bank Of India")
bank_1.info()
```

```python
user_1 = User("State Bank Of India", "Rohit")
user_1.user_info()
```

```python
user_1.info()
```

```python
# OOPs with Apple device 
# Parent Class - Hierarchical Inheritance
class Apple:
    def __init__(self, company_name):
        self.company_name = company_name
    
    def info(self):
        return f"Company: {self.company_name}"

# Child Class inheriting from Apple
class iPhone(Apple):
    def __init__(self, company_name, model):
        super().__init__(company_name)   # Call parent constructor
        self.model = model
    
    def iphone_info(self):
        return f"{self.company_nam}"

# Another Child Class inheriting from Apple
class iPad(Apple):
    def __init__(self, company_name, model):
        super().__init__(company_name)   # Call parent constructor
        self.model = model
    
    def ipad_info(self):
        return f"{self.company_name} - iPad Model: {self.model}"
```

```python
from abc import ABC , abstractmethod
```

```python
class BankServer(ABC) :
    def __init__ (self,bank_name) :
        self.bank_name = bank_name

    @abstractmethod
    def set_security(self) :
        pass

    def Card(BankServer) :
        def __init__ (self,bank_name,card_number) :
            super().__init__(bank_name)
            self.card_number = card_number

```

```python
# child class Online Banking

class OnlineBanking(BankServer) :
    def __init__(self , bank_name , user_id) :
        super().__init__(bank_name)
        self.user_id = user_id

    def set_security(self) :
        return f"Profile is secured with two factor authentication"
```

```python
from abc import ABC, abstractmethod
```

```python
# Abstract Parent Class
class Netflix(ABC):
    def __init__(self, company_name):
        self.company_name = company_name
    
    @abstractmethod
    def feature(self):
        pass
```

```python
# Child Class Streaming
class Streaming(Netflix):
    def __init__(self, company_name, show_name):
        super().__init__(company_name)
        self.show_name = show_name
    
    def feature(self):
        return f"{self.company_name}: Streaming '{self.show_name}' in HD"
```

```python
# Child Class Download
class Download(Netflix):
    def __init__(self, company_name, movie_name):
        super().__init__(company_name)
        self.movie_name = movie_name
    
    def feature(self):
        return f"{self.company_name}: Downloaded '{self.movie_name}' for offline viewing"
```

```python
# Child Class Recommendation
class Recommendation(Netflix):
    def __init__(self, company_name, user_name):
        super().__init__(company_name)
        self.user_name = user_name
    
    def feature(self):
        return f"{self.company_name}: Personalized recommendations for {self.user_name}"
```

```python
# Using objects
stream = Streaming("Netflix", "Stranger Things")
download = Download("Netflix", "Inception")
recommend = Recommendation("Netflix", "Rock")
```

```python
print(stream.feature())      
print(download.feature())
print(recommend.feature())
```

```python
class BankAccount :
    def __init__ (self , owner , balance ) :
        self.__owner = owner
        self.__balance = balance

    def get_owner(self) :
        return self.__owner

    def get_balance(self) :
        return self.__balance

    def deposit(self , amount) :
        self.__balance += amount

    def withdraw(self , amount) :
        if self.__balance >= amount :
           self.__balance -= amount
        else :
            return 

    def info(self) :
        return f" Account : {self.__owner} and the balance is {self.__balance}" 
    
```

```python
account1 =BankAccount("Rahul" , 10000)
```
