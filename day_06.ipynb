```python
# Demonstrating the difference between Tuple and List

# Creating a tuple and a list with the same elements
my_tuple = (1, 2, 3)
my_list = [1, 2, 3]

print("Original tuple:", my_tuple)
print("Original list:", my_list)

# Modifying the list
my_list.append(4)
print("List after append(4):", my_list)

my_list[0] = 100
print("List after changing first element:", my_list)

# Trying to modify the tuple (this will raise an error)
print("Attempting to modify the tuple...")
try:
    my_tuple[0] = 100
except TypeError as e:
    print("Error:", e)

# Tuple concatenation creates a new tuple
new_tuple = my_tuple + (4, 5)
print("Original tuple remains:", my_tuple)
print("New tuple after concatenation:", new_tuple)

# Checking hashability
print("Can tuple be a dictionary key? Yes")
sample_dict = {my_tuple: "This works"}
print("Dictionary with tuple key:", sample_dict)

print("Can list be a dictionary key? No")
try:
    bad_dict = {my_list: "This fails"}
except TypeError as e:
    print("Error:", e)

# Checking available methods
print("Tuple methods:", dir(my_tuple))
print("List methods include append, remove, sort, pop, insert, etc.")
```
