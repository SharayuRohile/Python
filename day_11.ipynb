```python
# Write a function for odd and even

def even_odd () :
    for i in range (1,11) :
        if i % 2 == 0 :
            print("Even")

        else :
            print("Odd")

even_odd()


```

```python
# Other way

def check_num (num) :
    if num % 2 == 0 :
        return "Even"
    else :
        return "Odd"

check_num(20)
```

```python
# Write a function for list to append 

def append_item(my_list, item):
    my_list.append(item)
    return my_list

numbers = [5, 2, 8, 2, 1]

numbers = append_item(numbers, 10)
print("After append:", numbers)
```

```python
# write a function to sort the list

def sort_list(my_list):
    return sorted(my_list)

numbers = sort_list(numbers)
print("Sorted list:", numbers)
```

```python
# write a function to check if  data contains duplicate or not

def has_duplicates(my_list):
    return len(my_list) != len(set(my_list))

print("Contains duplicates:", has_duplicates(numbers))
```
