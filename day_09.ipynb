```python
# Sets
# Duplicates not allowed

a={1,2,3,4,5,1,2}
print(a)
```

```python
# Add Method it will add only one element at a time
# sets don't follow the order

cart={"Milk","Rice","Vegetables"}
cart.add("Water")
print(cart)

cart={"Milk","Rice","Vegetables"}
cart.add("Water")
cart.add("Bowl")
print(cart)
```

```python
city={"Pune","Mumbai","Nashik"}
city.add("Nagpur")
print(city)
```

```python
record={"rahul","rahul","raj","rohan","virat","rohit"}

present=set()

for i in record :
    present.add(i)
print(present)
```

```python
# update Method 
# If we want to add more than one element at a time at that time we use update

fruits={"apple","jackfruit"}
fruits.update(["orange","mango","kiwi"])
print(fruits)
```

```python

# Remove Method
# When we want to remove element
# the item should present if not then error occurs

fruits={"apple","jackfruit","orange","mango"}
fruits.remove("apple")
print(fruits)


```

```python
# Discard method
# it will not give error if item doesn't exist in the set

fruits={"apple","jackfruit","orange","mango"}
fruits.discard("banana")
print(fruits)
```

```python
# Pop Method 
# randomly remove any item

lottery={"rock","john","big show","mark","goldi","brock"}
lottery.pop()
print(lottery)
```

```python
# Clear Method
# It will clear the items we will get empty set

lottery={"rock","john","big show","mark","goldi","brock"}
lottery.clear()
print(lottery)
```

```python
# Copy Method
# If we want to copy elements from one set to another we will use copy

a={"Mango","Kiwi"}
b=a.copy()
print(b)
```

```python
new_set=a.copy()
print(new_set)
print(a)
print(new_set)
```

```python
# Print Squares from set

nums={2,4,6,8}

for i in nums :
    print(i*i)
```

```python
# count elements
# use a for loop and no use of len()

elements = {1,2,3,4,5} 
len(elements)

# Another way

elements = {1,2,3,4,5} 
count=0 # empty space
for i in elements :
    count += 1
print(count)
```

```python
# Use for loop to find the percentage 


marks = [85, 67, 56, 45, 56, 67]
total = 0

for i in marks:
    total = total + i

percentage = total / 6

print("Total marks =", total)
print("Percentage =", percentage, "%")
```

```python
# Filter even numbers

nums = {1,2,3,4,5,6,7}

for i in nums :
    if i % 2 == 0 :
        print (i,"is even")
    else:
            print(i , "Number is odd")
```

```python
# find the max number without using max()

nums = [56, 67, 78, 456, 43, 56, 76, 8, 56]

largest = nums[0]

for i in nums:
    if i > largest:
        largest = i

print("Maximum number =", largest)
```
