```python
# break - when we get that element at that time it stops executing
# continue - it will skip for specific element and continue for others
# pass - it will skip that specific element and will pass to others
```

```python
marks=[45,65,67,23,54,67,89,45,35,67,89]
for i in marks:
    if i < 35:
        print(i,"you failed")
        break
    print(i)
```

```python
cart=["spinach","carrot","avacado","tomato"]
for i in cart:
    if i=="avacado":
        print("we found",i)
        break
    print(i)
```

```python
marks=[45,68,36,25,51,100]
for i in marks:
    if i < 35:
        print(i,"you failed the exam")
        break
    else:
        print(i,"you pass the exam")
```

```python
# continue - it will skip for oats and continue for other
snacks=["chips","tacco","burger","oats","pizza","pasta","softy"]
for i in snacks:
    if i=="oats":
        print("skip it please",i)
        continue
    print("Yummy Eating",i)
    
```

```python
# pass - 
marks=[78,65,85,67,56,100,87]
for i in marks:
    if i <  90:
        print("You are getting books",i)
    else:
        pass
```
